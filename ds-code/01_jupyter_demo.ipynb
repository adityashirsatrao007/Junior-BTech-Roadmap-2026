## Step 1 — load the Titanic starter dataset (from `data/titanic.csv`)


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/titanic.csv")   # path is relative to this notebook's folder
df.head()
```

```python
df.info()
print("---")
df.describe().T
```

## Step 2 — first insight: did class / sex decide survival?


```python
print(df.groupby("sex")["survived"].mean())
print(df.groupby("pclass")["survived"].mean().round(2))
```

```python
df.groupby("pclass")["survived"].mean().plot(kind="bar", title="Survival rate by class")
plt.ylabel("survival rate")
plt.grid(alpha=.3)
plt.show()
```
