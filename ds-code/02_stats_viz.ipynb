## 1. Load the two datasets


```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

houses  = pd.read_csv("../data/houses.csv")
titanic = pd.read_csv("../data/titanic.csv")
print("houses", houses.shape, "| titanic", titanic.shape)
```

## 2. Descriptive statistics — mean, median, std, quartiles


```python
print(houses[["SalePrice", "LotArea"]].describe().T)
```

## 3. Distribution + outliers — histogram and boxplot (see skew/outliers for real)


```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
sns.histplot(houses["SalePrice"], bins=40, kde=True, ax=ax[0])
ax[0].set_title("Price distribution (right-skewed)")
sns.boxplot(x=houses["LotArea"], ax=ax[1])
ax[1].set_title("LotArea boxplot (outliers visible)")
plt.tight_layout(); plt.show()
```

## 4. Correlation — the classic interview question "what drives price?"


```python
corr = houses.select_dtypes(include=np.number).corr()
print(corr["SalePrice"].sort_values(ascending=False))
```

```python
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation matrix (houses)")
plt.show()
```

## 5. Grouping — categorical vs numeric (crosstab you'll use forever)


```python
print(titanic.groupby("sex")["survived"].mean())
print("---")
print(pd.crosstab(titanic.pclass, titanic.sex, titanic.survived, aggfunc="mean"))
```
